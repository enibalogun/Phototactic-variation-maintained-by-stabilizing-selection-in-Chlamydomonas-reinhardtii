# Analysis

```python
#### List of sample names ####
CC2344 = ['CC2344-ANC', "CC2344-L1", "CC2344-L2", "CC2344-L3", "CC2344-L4", "CC2344-L5", "CC2344-L6", "CC2344-L7", "CC2344-L8", "CC2344-L9", "CC2344-L10", "CC2344-L11", "CC2344-L12", "CC2344-L13", "CC2344-L14", "CC2344-L15"]
CC2931 = ["CC2931-ANC", "CC2931-L1", "CC2931-L2", "CC2931-L3", "CC2931-L4", "CC2931-L5", "CC2931-L6", "CC2931-L7", "CC2931-L9", "CC2931-L10", "CC2931-L11", "CC2931-L13", "CC2931-L14", "CC2931-L15"]

#### Dataframe recording the generation time per sample ####
dic_gen = {}
dic_fitness = {}
generation = pd.read_csv('/research/projects/chlamydomonas/MAexpression/genome_info/mutation_info/Mutation_Fitness.txt', delimiter = '\t')
generation['Sample'] = generation['Sample'].str.replace('_', '-L')
generation = generation.loc[generation['Sample'].isin(CC2344[1:] + CC2931[1:])]

for i in generation.index.values:
    dic_gen[generation.at[i,'Sample']] = generation.at[i, 'Generation']
    dic_fitness[generation.at[i,'Sample']] = generation.at[i, 'Average_realtive']
    
generations = pd.Series(dic_gen)
generations['CC2344-ANC'] = 0
generations['CC2931-ANC'] = 0
generations = pd.DataFrame(generations, columns = ['generations']).reset_index()
generations.columns = ['line', 'generations']

#### Recording fitness per sample ####
fitness = pd.Series(dic_fitness)
fitness['CC2344-ANC'] = 0
fitness['CC2931-ANC'] = 0
fitness = pd.DataFrame(fitness, columns = ['fitness']).reset_index()
fitness.columns = ['line', 'fitness']

#### Dataframe recording the number of mutations per sample ####
all_mutations = pd.read_csv('/research/projects/chlamydomonas/MAexpression/genome_info/mutation_info/all_mutations.csv', delimiter = '\t')
dic_mut = {maline:all_mutations.loc[all_mutations['sample'] == maline].count()[0] for maline in all_mutations['sample'].values.tolist()}
mutations = pd.Series(dic_mut)
mutations['CC2344-ANC'] = 0
mutations['CC2931-ANC'] = 0
mutations = pd.DataFrame(mutations, columns = ['mutations']).reset_index()
mutations.columns = ['line', 'mutations']
```

## Genetic Relatedness Matrix

```python
grm_file = '/research/projects/chlamydomonas/NIexpression/data/snp_differences/kinship_grm.grm.gz'
dist = pd.read_csv(grm_file, delimiter = ',', header = None)
dist = dist[0].str.split(' ', expand = True)
dist.columns = ['ID1', 'ID2', 'shared_SNPs', 'grm']

sample_id = pd.read_csv('/research/projects/chlamydomonas/NIexpression/data/snp_differences/kinship_grm.grm.id', delimiter = ' ', header = None)
dic = {i+1:sample_id.loc[i, 1] for i in range(len(sample_id))}
dist['ID1'] = dist['ID1'].astype(int)
dist['ID2'] = dist['ID2'].astype(int)
dist['shared_SNPs'] = dist['shared_SNPs'].astype(int)
dist['grm'] = dist['grm'].astype(float)
dist['ID1'] = dist['ID1'].replace(dic)
dist['ID2'] = dist['ID2'].replace(dic)

NI = ['GB123', 'GB117', 'GB66', 'GB49', 'CC3059', 'CC3060', 'CC-4532', 'W13-1', 'W13-2', 'CC2344', 'CC2931', 'JG4-1', 'CL3-1', 'CC3079']

dist = pd.concat([dist.loc[(dist['ID1'].isin(NI)) & (dist['ID2'] == 'GB123')], dist.loc[(dist['ID2'].isin(NI)) & (dist['ID1'] == 'GB123')]]).drop_duplicates()
dist
```

```python
mota = pd.read_csv('/research/projects/chlamydomonas/PhototacticRate/data/motility_assay.csv', delimiter = '\t')[['strain', 'distance', 'variable', 'value', 'type']]
mota = mota.loc[mota['distance'] == 'Maximum Distance']
mota['strain'] = mota['strain'].str.replace('-', '').str.replace(' ', '-')
mota = mota.loc[mota['type'] == 'NI']
mota['average_distance'] = 'NaN'
strains = mota['strain'].unique()

dic = {}
for i in strains:
    dic[i] = mota.loc[mota['strain'] == i]['value'].mean()

df = pd.DataFrame.from_dict(dic, orient='index', columns=['average_maxdistance'])
df.index = ['JG4-1', 'GB66', 'GB49', 'W13-1', 'W13-2', 'GB123', 'GB117', 'CC3059', 'CC4532', 'CC3079', 'CC3060', 'CL3-1']
for i in df.index:
    try:
        df.loc[i, 'grm'] = dist.loc[(dist['ID1'] == i) | (dist['ID2'] == i)]['grm'].values[0]
    except: continue
df = df.dropna()
sns.scatterplot(df, x = 'grm', y = 'average_maxdistance')
```

## Statistical Analysis

### Average Maximum Distance Analysis

```python
mota = pd.read_csv('/research/projects/chlamydomonas/PhototacticRate/data/motility_assay_3.csv', delimiter = '\t')[['strain', 'distance', 'variable', 'value', 'type']]
mota = mota.loc[mota['distance'] == 'Maximum Distance']
mota['strain'] = mota['strain'].str.replace('-', '').str.replace(' ', '-')
mota = mota.loc[mota['type'] != 'NI']
mota.columns = ['line', 'distance', 'variable', 'value', 'type']
mota = pd.merge(mota, mutations, on = 'line')
mota = pd.merge(mota, generations, on = 'line')
mota = pd.merge(mota, fitness, on = 'line')

CC2344_ANC = mota.loc[mota['type'] == 'CC-2344 ANC']['value'].mean()
CC2931_ANC = mota.loc[mota['type'] == 'CC-2931 ANC']['value'].mean()

mota.loc[mota.loc[mota['type'] == 'CC-2344 MA'].index, 'relative_maxdistance'] = mota.loc[mota['type'] == 'CC-2344 MA']['value']/CC2344_ANC
mota.loc[mota.loc[mota['type'] == 'CC-2931 MA'].index, 'relative_maxdistance'] = mota.loc[mota['type'] == 'CC-2931 MA']['value']/CC2931_ANC

for strain in mota['line'].unique():
    avg = mota.loc[mota['line'] == strain]
    mota.loc[avg.index, 'average_maxdistance'] = avg['value'].mean()
```

```python
mota_subset = mota.drop_duplicates(subset = ['line'])

#### Correlation with mutations in all lines ####
model = smf.ols('average_maxdistance ~ mutations', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with mutations in all lines (excluding mutators) ####
model = smf.ols('average_maxdistance ~ mutations', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])].loc[mota_subset['mutations'] < 300]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with mutations + generations ####
model = smf.ols('average_maxdistance ~ mutations + generations', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with mutations + generations + interaction term ####
model = smf.ols('average_maxdistance ~ mutations + generations + mutations*generations', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with fitness ####
model = smf.ols('average_maxdistance ~ fitness', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with mutations + fitness ####
model = smf.ols('average_maxdistance ~ mutations + fitness', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with mutations + fitness + interaction term ####
model = smf.ols('average_maxdistance ~ mutations + fitness + mutations*fitness', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with generations ####
model = smf.ols('average_maxdistance ~ generations', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with generations + fitness ####
model = smf.ols('average_maxdistance ~ generations + fitness', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)

#### Correlation with generations + fitness + interaction term ####
model = smf.ols('average_maxdistance ~ generations + fitness + generations*fitness', data=mota_subset.loc[~mota_subset['type'].isin(['CC-2344 ANC', 'CC-2931 ANC'])]).fit()
display(model.summary())
anova_table = anova_lm(model)
display(anova_table)
```

### Average Relative Maximum Distance Analysis

```python
mota = pd.read_csv('/research/projects/chlamydomonas/PhototacticRate/data/motility_assay_3.csv', delimiter = '\t')[['strain', 'distance', 'variable', 'value', 'type']]
mota = mota.loc[mota['distance'] == 'Maximum Distance']
mota['strain'] = mota['strain'].str.replace('-', '').str.replace(' ', '-')
mota = mota.loc[mota['type'] != 'NI']
mota.columns = ['line', 'distance', 'variable', 'value', 'type']
mota = pd.merge(mota, mutations, on = 'line')
mota = pd.merge(mota, generations, on = 'line')
mota = pd.merge(mota, fitness, on = 'line')

CC2344_ANC = mota.loc[mota['type'] == 'CC-2344 ANC']['value'].mean()
CC2931_ANC = mota.loc[mota['type'] == 'CC-2931 ANC']['value'].mean()

#### Average phototactic rate of each strain ####
for strain in mota['line'].unique():
    avg = mota.loc[mota['line'] == strain]
    mota.loc[avg.index, 'average_maxdistance'] = avg['value'].mean()

mota.loc[mota.loc[mota['type'] == 'CC-2344 MA'].index, 'relative_average_maxdistance'] = mota.loc[mota['type'] == 'CC-2344 MA']['average_maxdistance']/CC2344_ANC
mota.loc[mota.loc[mota['type'] == 'CC-2931 MA'].index, 'relative_average_maxdistance'] = mota.loc[mota['type'] == 'CC-2931 MA']['average_maxdistance']/CC2931_ANC

mota['type'] = mota['type'].str.split(' ', expand = True)[0]

mota = mota.loc[~mota['line'].isin(['CC2344-ANC', 'CC2931-ANC'])]
mota = mota.drop_duplicates(subset = ['line'], keep = 'first')
```

### Plot - Mutation Count vs. Relative Distance

```python
%%R

# packageurl<-"https://cran.r-project.org/src/contrib/Archive/nloptr/nloptr_1.2.1.tar.gz"
# install.packages(packageurl, repos=NULL, type="source")

# install.packages("Matrix")
# install.packages("lme4")
# install.packages('insight')

library(insight)
library(Matrix)
library(lme4)
```

## CC-2931

```python
%%R

#### CC-2931 ####
#################

CC2931_data <- read.csv('/research/projects/chlamydomonas/PhototacticRate/data/CC2931_combined_gene_info_mota3', sep = '\t', header = TRUE, stringsAsFactors = FALSE)

CC2931_noANC <- CC2931_data[CC2931_data$line != 'CC2931-ANC', ]
CC2931_noANC$obs <- 1:nrow(CC2931_noANC) ## assigning obs for each unique observation to add dispersion
results <-matrix(NA, 4, 4)

### Mutational Variance ####
############################
CC2931_noANC$obs <- as.factor(CC2931_noANC$obs)

m1 <- lmer(maxdistance ~ generations -1 + (sqrt(generations) -1|line) + (1|rep), data = CC2931_noANC)

m2 <- lmer(maxdistance ~ mutations -1 + (sqrt(mutations) -1|line) + (1|rep), data = CC2931_noANC)

CC2931_noANC <- CC2931_noANC[CC2931_noANC$mutations < 200,] ## ignoring genes with possible mutators

m3 <- lmer(maxdistance ~ generations -1 + (sqrt(generations) -1|line) + (1|rep), data = CC2931_noANC)

m4 <- lmer(maxdistance ~ mutations -1 + (sqrt(mutations) -1|line) + (1|rep), data = CC2931_noANC)

results[1,1] <- VarCorr(m1)$line[1]                # between-line variance estimate
results[1,2] <- VarCorr(m2)$line[1]
results[1,3] <- VarCorr(m3)$line[1]                # between-line variance estimate
results[1,4] <- VarCorr(m4)$line[1]

results[2,1] <- get_variance_residual(m1)
results[2,2] <- get_variance_residual(m2)
results[2,3] <- get_variance_residual(m3)
results[2,4] <- get_variance_residual(m4)
```
